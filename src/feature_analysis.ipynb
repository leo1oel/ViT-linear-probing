```python
import os
import h5py
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
from sklearn.metrics import pairwise_distances
import umap

def load_features(file_path):
    """
    从 h5 文件中加载特征与标签
    """
    with h5py.File(file_path, 'r') as f:
        features = f['last_hidden_cls'][:]
        labels = f['targets'][:]
    return features, labels

def filter_classes(features, labels, selected_classes):
    """
    只保留标签属于 selected_classes 的样本
    """
    mask = np.isin(labels, selected_classes)
    return features[mask], labels[mask]

def compute_similarity_metrics(features, labels, metric='cosine'):
    """
    计算类内与类间相似性。使用给定距离度量（默认余弦距离）计算所有样本间的相似度，
    然后统计同类样本与不同类样本的平均相似度。
    """
    dists = pairwise_distances(features, metric=metric)
    similarities = 1 - dists  # 转换为相似度
    unique_labels = np.unique(labels)
    intra_list = []
    inter_list = []
    for lbl in unique_labels:
        idx = np.where(labels == lbl)[0]
        if len(idx) > 1:
            sim = similarities[np.ix_(idx, idx)]
            intra = sim[np.triu_indices_from(sim, k=1)]
            intra_list.append(intra)
        # 计算当前类别与其他类别样本间的相似度
        other_idx = np.where(labels != lbl)[0]
        if len(idx) > 0 and len(other_idx) > 0:
            inter_sim = similarities[np.ix_(idx, other_idx)]
            inter_list.append(inter_sim.flatten())
    intra_sim = np.concatenate(intra_list) if intra_list else np.array([])
    inter_sim = np.concatenate(inter_list) if inter_list else np.array([])
    return np.mean(intra_sim), np.mean(inter_sim)

def tsne_plot(features, labels, title, sample_size=50000, alpha=0.6):
    """
    使用 t-SNE 对样本进行降维，可选对样本数目进行采样以加快计算速度
    """
    if features.shape[0] > sample_size:
        indices = np.random.choice(features.shape[0], sample_size, replace=False)
        features_sample = features[indices]
        labels_sample = labels[indices]
    else:
        features_sample = features
        labels_sample = labels

    tsne = TSNE(n_components=2, random_state=42, init='pca')
    features_2d = tsne.fit_transform(features_sample)
    plt.figure(figsize=(8, 6))
    scatter = plt.scatter(features_2d[:, 0], features_2d[:, 1],
                          c=labels_sample, cmap='tab10', s=10, alpha=alpha)
    plt.title(title)
    plt.xlabel("t-SNE 维度 1")
    plt.ylabel("t-SNE 维度 2")
    plt.colorbar(scatter, label='类别')
    plt.tight_layout()
    plt.show()

def umap_plot(features, labels, title, sample_size=2000, alpha=0.6):
    """
    使用 UMAP 对样本进行降维，可选采样以加快计算速度
    """
    if features.shape[0] > sample_size:
        indices = np.random.choice(features.shape[0], sample_size, replace=False)
        features_sample = features[indices]
        labels_sample = labels[indices]
    else:
        features_sample = features
        labels_sample = labels

    reducer = umap.UMAP(random_state=42)
    features_2d = reducer.fit_transform(features_sample)
    plt.figure(figsize=(8, 6))
    scatter = plt.scatter(features_2d[:, 0], features_2d[:, 1],
                          c=labels_sample, cmap='tab10', s=10, alpha=alpha)
    plt.title(title)
    plt.xlabel("UMAP 维度 1")
    plt.ylabel("UMAP 维度 2")
    plt.colorbar(scatter, label='类别')
    plt.tight_layout()
    plt.show()

def analyze_dataset(model_name, dataset_name, file_path, selected_classes=None):
    """
    对单个数据集与模型，加载数据：
      - 根据 selected_classes 参数过滤样本（如果不为 None）
      - 计算类内与类间相似性
      - 分别使用 t-SNE 和 UMAP 进行降维可视化
    """
    print(f"Analyzing {model_name} on {dataset_name} ...")
    features, labels = load_features(file_path)
    if selected_classes is not None:
        features, labels = filter_classes(features, labels, selected_classes)
        print(f"只分析类别: {selected_classes}, 总样本数: {features.shape[0]}")

    intra, inter = compute_similarity_metrics(features, labels, metric='cosine')
    ratio = intra / inter if inter != 0 else np.nan
    print(f"Results for {model_name} on {dataset_name}:")
    print(f"  - Mean intra-class similarity: {intra:.4f}")
    print(f"  - Mean inter-class similarity: {inter:.4f}")
    print(f"  - Intra/Inter similarity ratio: {ratio:.4f}\n")

    tsne_plot(features, labels, title=f"{model_name} on {dataset_name} t-SNE")
    umap_plot(features, labels, title=f"{model_name} on {dataset_name} UMAP")

def main():
    # 根据数据集和模型组织文件路径字典，请将下面的路径替换为实际文件路径
    data_files = {
        # 'imagenet': {
        #     'clip': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/imagenet/clip-datacomp_val_features.h5',
        #     'dino': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/imagenet/dino-datacomp_val_features.h5'
        # },
        # 'cifar10': {
        #     'clip': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/cifar10/clip-datacomp_train_features.h5',
        #     'dino': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/cifar10/dino-datacomp_train_features.h5'
        # },
        'stanford_flowers': {
            'clip': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/flowers102/clip-datacomp_train_features.h5',
            'dino': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/flowers102/dino-datacomp_train_features.h5'
        },
        'stanford_cars': {
            'clip': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/stanford_cars/clip-datacomp_train_features.h5',
            'dino': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/stanford_cars/dino-datacomp_train_features.h5'
        },
        'CUB': {
            'clip': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/cub/clip-datacomp_train_features.h5',
            'dino': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/cub/dino-datacomp_train_features.h5'
        }
    }
    
    # 举例：若类别过多，可以只选一些重点类别进行分析，例如只关注类别 0, 1, 2
    selected_classes = list(range(21))  # 根据需要修改类别号

    # 分析各数据集下的 CLIP 和 DINO
    for dataset, models in data_files.items():
        for model_name, file_path in models.items():
            if os.path.exists(file_path):
                # 如果你希望全部类别分析，将 selected_classes 设置为 None
                analyze_dataset(model_name, dataset, file_path, selected_classes)
            else:
                print(f"文件 {file_path} 不存在，请检查 {model_name} 在 {dataset} 上的文件路径。")

if __name__ == '__main__':
    main()
```

```python
import os
import h5py
import numpy as np
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE
from sklearn.metrics import pairwise_distances

def load_features(file_path):
    """
    从给定 h5 文件中加载特征和标签
    """
    with h5py.File(file_path, 'r') as f:
        features = f['last_hidden_cls'][:]
        labels = f['targets'][:]
    return features, labels

def compute_similarity_metrics(features, labels, metric='cosine'):
    """
    计算类内与类间余弦相似性均值。
    类内相似性——同一类别中任意两个样本之间的相似度均值；
    类间相似性——不同类别中任意两个样本之间的相似度均值。
    """
    # pairwise_distances 返回的是距离，若使用余弦距离，则相似度可以表示为 1 - 距离
    dists = pairwise_distances(features, metric=metric)
    similarities = 1 - dists
    unique_labels = np.unique(labels)
    intra_list = []
    inter_list = []
    for lbl in unique_labels:
        idx = np.where(labels == lbl)[0]
        # 计算类内相似性（去除对角线上的自相似性）
        if len(idx) > 1:
            sim = similarities[np.ix_(idx, idx)]
            intra = sim[np.triu_indices_from(sim, k=1)]
            intra_list.append(intra)
        # 计算类间相似性：当前类别的样本与其他类别样本间的相似度
        other_idx = np.where(labels != lbl)[0]
        if len(idx) > 0 and len(other_idx) > 0:
            inter_sim = similarities[np.ix_(idx, other_idx)]
            inter_list.append(inter_sim.flatten())
    intra_sim = np.concatenate(intra_list) if intra_list else np.array([])
    inter_sim = np.concatenate(inter_list) if inter_list else np.array([])
    return np.mean(intra_sim), np.mean(inter_sim)

def tsne_plot(features, labels, title, sample_size=10000):
    """
    对特征进行 t-SNE 降维并绘制散点图。
    若样本数目过多，则随机采样 sample_size 个样本进行 t-SNE 加速。
    """
    if features.shape[0] > sample_size:
        indices = np.random.choice(features.shape[0], sample_size, replace=False)
        features_sample = features[indices]
        labels_sample = labels[indices]
    else:
        features_sample = features
        labels_sample = labels
    tsne = TSNE(n_components=2, random_state=42, init='pca')
    features_2d = tsne.fit_transform(features_sample)
    plt.figure(figsize=(8, 6))
    scatter = plt.scatter(features_2d[:, 0], features_2d[:, 1], c=labels_sample, cmap='tab10', s=10, alpha=0.3)
    plt.title(title)
    plt.xlabel("t-SNE 维度 1")
    plt.ylabel("t-SNE 维度 2")
    plt.colorbar(scatter, label='类别')
    plt.tight_layout()
    plt.show()

def analyze_dataset(model_name, dataset_name, file_path):
    """
    对单个数据集的单个模型进行分析，输出类内和类间相似性指标，并展示 t-SNE 可视化结果
    """
    print(f"Analyzing {model_name} on {dataset_name} ...")
    features, labels = load_features(file_path)
    intra, inter = compute_similarity_metrics(features, labels, metric='cosine')
    ratio = intra / inter if inter != 0 else np.nan
    print(f"Results for {model_name} on {dataset_name}:")
    print(f"  - Mean intra-class similarity: {intra:.4f}")
    print(f"  - Mean inter-class similarity: {inter:.4f}")
    print(f"  - Intra/Inter similarity ratio: {ratio:.4f}\n")
    tsne_plot(features, labels, title=f"{model_name} on {dataset_name} t-SNE Visualization")

def main():
    # 根据数据集和模型组织文件路径字典，请将下面的路径修改为你的实际文件路径
    data_files = {
        # 'imagenet': {
        #     'clip': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/imagenet/clip-datacomp_val_features.h5',
        #     'dino': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/imagenet/dino-datacomp_val_features.h5'
        # },
        # 'cifar10': {
        #     'clip': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/cifar10/clip-datacomp_train_features.h5',
        #     'dino': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/cifar10/dino-datacomp_train_features.h5'
        # },
        'stanford_flowers': {
            'clip': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/flowers102/clip-datacomp_train_features.h5',
            'dino': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/flowers102/dino-datacomp_train_features.h5'
        },
        'stanford_cars': {
            'clip': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/stanford_cars/clip-datacomp_train_features.h5',
            'dino': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/stanford_cars/dino-datacomp_train_features.h5'
        },
        'CUB': {
            'clip': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/cub/clip-datacomp_train_features.h5',
            'dino': '/pasteur2/u/yuhuiz/yiming/experiments/src/cached_features/cub/dino-datacomp_train_features.h5'
        }
    }

    # 分别对各数据集下的 CLIP 和 DINO 进行分析
    for dataset, models in data_files.items():
        for model_name, file_path in models.items():
            if os.path.exists(file_path):
                analyze_dataset(model_name, dataset, file_path)
            else:
                print(f"文件 {file_path} 不存在，请检查 {model_name} 在 {dataset} 上的文件路径。")

if __name__ == '__main__':
    main()
```
